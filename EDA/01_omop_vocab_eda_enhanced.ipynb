# OMOP Vocabulary EDA: Complete Analysis of All Files

This notebook provides comprehensive analysis of ALL OMOP Vocabulary files including:
- CONCEPT, VOCABULARY, DOMAIN, RELATIONSHIP
- DRUG_STRENGTH, CONCEPT_RELATIONSHIP, CONCEPT_SYNONYM
- CONCEPT_CLASS, CONCEPT_ANCESTOR, CONCEPT_CPT4

Each file includes df.head(5) display, statistics, and visualizations.

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from IPython.display import display, Markdown
import warnings
warnings.filterwarnings('ignore')

# Settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
plt.style.use('bmh')
sns.set_palette('husl')

data_path = r'D:\ADE DATASET DOWNLOAD\omop vocab'
eda_path = r'D:\ADE DATASET DOWNLOAD\EDA'

print(f"OMOP Vocabulary Directory: {data_path}")
print(f"Directory exists: {os.path.exists(data_path)}")
```

## 1. File Inventory & Size Analysis

```python
files = [f for f in os.listdir(data_path) if f.endswith('.csv')]
file_stats = []
for f in files:
    path = os.path.join(data_path, f)
    size_mb = os.path.getsize(path) / (1024 * 1024)
    file_stats.append({'File': f, 'Size (MB)': round(size_mb, 2)})

df_file_stats = pd.DataFrame(file_stats).sort_values('Size (MB)', ascending=False)
display(Markdown("### OMOP Vocabulary File Inventory"))
display(df_file_stats)

# Visualization
plt.figure(figsize=(12, 6))
sns.barplot(data=df_file_stats, x='Size (MB)', y='File', palette='viridis')
plt.title('OMOP Vocabulary Files by Size')
plt.xlabel('Size (MB)')
plt.tight_layout()
plt.show()
```

## 2. CONCEPT Table - Core Vocabulary

```python
display(Markdown("### Loading CONCEPT.csv"))
concept = pd.read_csv(os.path.join(data_path, 'CONCEPT.csv'), sep='\t', low_memory=False)

display(Markdown(f"**Total Concepts:** {len(concept):,}"))
display(Markdown("#### First 5 Rows:"))
display(concept.head(5))

display(Markdown("#### Data Types & Info:"))
print(concept.info())

display(Markdown("#### Statistical Summary:"))
display(concept.describe(include='all'))
```

```python
# Concept Distributions
fig, axes = plt.subplots(2, 2, figsize=(20, 14))

# Top 20 Vocabularies
vocab_counts = concept['vocabulary_id'].value_counts().head(20)
sns.barplot(x=vocab_counts.values, y=vocab_counts.index, ax=axes[0,0], palette='viridis')
axes[0,0].set_title('Top 20 Vocabularies by Concept Count', fontsize=14, fontweight='bold')
axes[0,0].set_xlabel('Number of Concepts')

# Top 20 Domains
domain_counts = concept['domain_id'].value_counts().head(20)
sns.barplot(x=domain_counts.values, y=domain_counts.index, ax=axes[0,1], palette='magma')
axes[0,1].set_title('Top 20 Domains by Concept Count', fontsize=14, fontweight='bold')
axes[0,1].set_xlabel('Number of Concepts')

# Concept Classes
class_counts = concept['concept_class_id'].value_counts().head(20)
sns.barplot(x=class_counts.values, y=class_counts.index, ax=axes[1,0], palette='crest')
axes[1,0].set_title('Top 20 Concept Classes', fontsize=14, fontweight='bold')
axes[1,0].set_xlabel('Number of Concepts')

# Standard Concept Distribution
std_concept = concept['standard_concept'].value_counts()
axes[1,1].pie(std_concept.values, labels=std_concept.index, autopct='%1.1f%%', startangle=90)
axes[1,1].set_title('Standard Concept Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 10. CONCEPT_ANCESTOR Table - Hierarchy Analysis

## 11. Domain vs Class Heatmap - Standard Concepts

```python
standard_concepts = concept[concept['standard_concept'] == 'S']
heatmap_data = standard_concepts.groupby(['domain_id', 'concept_class_id']).size().unstack(fill_value=0)

# Filter for top domains and classes
top_domains = standard_concepts['domain_id'].value_counts().head(15).index
top_classes = standard_concepts['concept_class_id'].value_counts().head(15).index
heatmap_subset = heatmap_data.loc[top_domains, top_classes]

plt.figure(figsize=(16, 10))
sns.heatmap(heatmap_subset, annot=True, fmt='d', cmap='YlGnBu', cbar_kws={'label': 'Concept Count'})
plt.title('Standard Concept Distribution: Domain vs Class (Top 15)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 12. Data Quality: Invalid & Deprecated Concepts

```python
invalid_stats = concept.groupby(['vocabulary_id', 'invalid_reason']).size().unstack(fill_value=0)
invalid_stats['Total Invalid'] = invalid_stats.sum(axis=1)
invalid_summary = invalid_stats.sort_values('Total Invalid', ascending=False).head(20)

display(Markdown("### Top 20 Vocabularies with Deprecated/Invalid Concepts"))
display(invalid_summary)

plt.figure(figsize=(12, 6))
invalid_summary['Total Invalid'].plot(kind='bar', color='salmon')
plt.title('Total Invalid Concepts by Vocabulary (Top 20)', fontsize=14, fontweight='bold')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 15. Cross-Vocabulary Mapping Summary

```python
# Analyze mapping between vocabularies
key_vocabs = ['SNOMED', 'RxNorm', 'ICD10CM', 'ICD9CM', 'LOINC', 'NDC', 'CPT4']
mapping_summary = []

for vocab in key_vocabs:
    total = (concept['vocabulary_id'] == vocab).sum()
    standard = ((concept['vocabulary_id'] == vocab) & (concept['standard_concept'] == 'S')).sum()
    mapping_summary.append({
        'Vocabulary': vocab,
        'Total Concepts': total,
        'Standard Concepts': standard,
        'Pct Standard': f"{(standard/total*100) if total > 0 else 0:.1f}%"
    })

mapping_df = pd.DataFrame(mapping_summary)
display(Markdown("### Key Vocabularies Mapping Summary"))
display(mapping_df)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(data=mapping_df, x='Total Concepts', y='Vocabulary', ax=axes[0], palette='mako')
axes[0].set_title('Total Concepts by Key Vocabulary', fontsize=12, fontweight='bold')

sns.barplot(data=mapping_df, x='Standard Concepts', y='Vocabulary', ax=axes[1], palette='rocket')
axes[1].set_title('Standard Concepts by Key Vocabulary', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 16. Final Coverage Report

```python
coverage_report = concept.groupby(['vocabulary_id', 'domain_id', 'standard_concept']).size().reset_index(name='concept_count')
report_path = os.path.join(eda_path, 'omop_concept_coverage_report.csv')
coverage_report.to_csv(report_path, index=False)

display(Markdown(f"### Coverage Report Saved to: `{report_path}`"))
display(Markdown("#### Top 50 Combinations:"))
display(coverage_report.sort_values('concept_count', ascending=False).head(50))

print(f"\n✓ Successfully analyzed all OMOP Vocabulary files!")
print(f"✓ Total files processed: {len(files)}")
print(f"✓ Total concepts analyzed: {len(concept):,}")
```
